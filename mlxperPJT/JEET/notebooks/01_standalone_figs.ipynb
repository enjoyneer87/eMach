# 01 · 자립 그림 (입력 데이터 없음)

해석 결과나 아카이브 없이 **코드만으로** 그려지는 그림.
어느 머신에서든 수 초에 끝난다.

| 그림 | 내용 | 스크립트 |
|---|---|---|
| Fig 7 | 제안 프레임워크 워크플로 도식 | `run_workflow_fig.py` |
| Fig A.1 | 표피 인자 · 근접 커널 $\eta K(\eta)$ | `run_figA1_eddy_factors.py` |

> **주의** 이 두 그림은 수치를 그림 안에 문자열로 박는다. 본문 수치가 바뀌면 자동으로 따라오지 않는다 — 노트북 02 의 감사 셀이 이를 잡는다.


```python
import os, sys, runpy, time

# 배치 불문 부트스트랩 — 워크트리(mlxperPJT/JEET)와 배포 패키지 양쪽에서 돈다.
ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
SCRIPTS = os.path.join(ROOT, 'scripts')
if not os.path.isdir(SCRIPTS):
    SCRIPTS = ROOT
LIB = ROOT if os.path.isdir(os.path.join(ROOT, 'jeet_acloss_rbf')) \
    else os.path.abspath(os.path.join(ROOT, '..', '..', 'tools'))
for p in (LIB, SCRIPTS):
    if p not in sys.path:
        sys.path.insert(0, p)
DATA = os.path.join(ROOT, 'data', 'e10')
if not os.path.isdir(DATA):
    DATA = os.path.join(ROOT, 'map_exports', 'e10')
os.environ.setdefault('JEET_DATA_ROOT', DATA)
# 그림 출력 폴더 — 기본은 fig_out/ (원고 fig/ 로 보내려면 JEET_FIGDIR)
os.environ.setdefault('JEET_FIGDIR', os.path.join(ROOT, 'fig_out'))
os.makedirs(os.environ['JEET_FIGDIR'], exist_ok=True)
print('ROOT   :', ROOT)
print('SCRIPTS:', SCRIPTS)
print('DATA   :', os.environ['JEET_DATA_ROOT'])
print('FIGDIR :', os.environ['JEET_FIGDIR'])
```

```python
def run(script, note=''):
    """run_*.py 를 __main__ 으로 실행하고 소요 시간을 찍는다."""
    p = os.path.join(SCRIPTS, script)
    assert os.path.exists(p), p
    t0 = time.time()
    print('>>>', script, note)
    try:
        runpy.run_path(p, run_name='__main__')
    except SystemExit as e:            # raise SystemExit(main()) 관용구
        if e.code:
            raise
    print('    %.1f s' % (time.time() - t0))
```

## Fig 7 — 워크플로 도식


```python
run('run_workflow_fig.py', '(Fig 7)')
```

## Fig A.1 — 와전류 인자


```python
run('run_figA1_eddy_factors.py', '(Fig A.1)')
```
