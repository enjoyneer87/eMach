# 03 · 필드 아카이브 기반 그림 (로컬 전용)

`map_exports/e10/fields/` 의 전주기 필드 텍스트(**약 2.6 GB**)가 필요하다. 배포 레포에는 포함되지 않으므로 이 노트북은 **원본 해석 환경에서만** 돈다.

| 그림 | 내용 | 스크립트 | 대략 소요 |
|---|---|---|---|
| Fig 1 | 도체 내 $J_{e,rms}$ TS vs 2-D 재구성 | `run_fig1_shared_scale.py` | 모델당 ~2 분 |
| Fig 3 | 상사 대응 운전점 요소별 필드 | `run_fig11_mvp_field.py` | 수 분 |
| Fig B.1 | 고정 필드 원천의 평가 변형 비교 | `run_fig_hybrid_variants.py` | 수 분 |

> Fig 1 은 두 모델 공통 색 스케일(`SHARED_VLIM`)과 반경 눈금[mm]을 쓴다 — 눈금 간격이 같아야 슬롯 높이 비교가 성립한다.


```python
import os, sys, runpy, time

# 이 노트북이 있는 곳의 부모 = mlxperPJT/JEET
JEET = os.path.abspath(os.path.join(os.getcwd(), '..'))
TOOLS = os.path.abspath(os.path.join(JEET, '..', '..', 'tools'))
if TOOLS not in sys.path:
    sys.path.insert(0, TOOLS)

# 그림 출력 폴더 — 배포 레포에서는 여기만 바꾸면 된다
os.environ.setdefault('JEET_FIGDIR',
                      os.path.abspath(os.path.join(JEET, 'fig_out')))
os.makedirs(os.environ['JEET_FIGDIR'], exist_ok=True)
print('JEET  :', JEET)
print('FIGDIR:', os.environ['JEET_FIGDIR'])
```

```python
def run(script, note=''):
    """run_*.py 를 __main__ 으로 실행하고 소요 시간을 찍는다."""
    p = os.path.join(JEET, script)
    assert os.path.exists(p), p
    t0 = time.time()
    print('>>>', script, note)
    runpy.run_path(p, run_name='__main__')
    print('    %.1f s' % (time.time() - t0))
```

## 아카이브 존재 확인


```python
fields = os.path.join(JEET, 'map_exports', 'e10', 'fields')
if os.path.isdir(fields):
    n = sum(len(f) for _, _, f in os.walk(fields))
    print('필드 아카이브 있음 — 파일 %d개' % n)
else:
    print('[!] 필드 아카이브 없음 — 이 노트북은 실행 불가')
```

## Fig 1 — 전류 집중 (TS vs 2-D)
