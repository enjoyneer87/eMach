```python



# Disable auto-save while running vbs script in Maxwell to speed up process.

autoSaveConfig = oDesktop.GetAutoSaveEnabled()

oDesktop.EnableAutoSave(0)



# Check version so correct project file extension is used
```

```python
versionArray = oDesktop.GetVersion().split(".")
projExt = ".aedt"
```

```python





oDesktop.RestoreWindow()

oProject = oDesktop.NewProject()

oProject.InsertDesign("Maxwell 2D", "Motor-CAD 251114_C67_test_Moa_copy2", "Transient", "")

oDefinitionManager = oProject.GetDefinitionManager()

oMaterialMgr = oDefinitionManager.GetManager("Material")

oDesign = oProject.SetActiveDesign("Motor-CAD 251114_C67_test_Moa_copy2")

ModelName = "251114_C67_test_Moa_copy2"
ModelName_ANSYSEM = "251114_C67_test_Moa_copy2_ANSYSEM_2D"
designName = "Motor-CAD" + ModelName
```

```python

# Add timestamp to model name if project with same name already exists

projects = oDesktop.GetProjects()
```

```python


import os

from datetime import datetime
for project in projects:

  project_name = project.GetName()

  if project_name == ModelName_ANSYSEM:

     # Replace slashes and colons in timestamp so only valid characters used for project name

    currentTime = str(datetime.now())

    currentTime = currentTime.replace(":", ".")

    currentTime = currentTime.replace("/", "-")

    currentTime = currentTime.replace(" ", "_")

    ModelName_ANSYSEM = ModelName_ANSYSEM + "_" + currentTime

    break

```

```python


oProject.SaveAs("E:\\KDH\\KJS\\" + ModelName_ANSYSEM + projExt, True)

```




# Ansys script methods


```python

def ChangePropertyName(name, name_New):

  oEditor.ChangeProperty(["NAME:AllTabs",

    ["NAME:Geometry3DAttributeTab",

    ["NAME:PropServers", str(name)],

    ["NAME:ChangedProps",

    ["NAME:Name", "Value:=", str(name_New)]]]])
```

```python



def SubtractObjects(blankPartList, toolPartList, keepOriginals):

  # Subtract tool parts boundaries from blank part boundaries

  oEditor.Subtract(["NAME:Selections",

    "Blank Parts:=", str(blankPartList),

    "Tool Parts:=", str(toolPartList)],

    ["NAME:SubtractParameters", "KeepOriginals:=", keepOriginals])

```

```python


def CreateOutputVariable(variableName, Expression):

  oOutputVarModule.CreateOutputVariable(str(variableName),

    str(Expression),

    "Setup1 : Transient", "Transient", [])



def BandToolSubtraction(bandTool):

  numObjects = oEditor.GetNumObjects()

  objectString = ""

  for objectLoop in range(numObjects - 1):

    name = oEditor.GetObjectName(int(objectLoop))

    if bandTool != str(name):

      objectString = objectString + str(name) + ","



  oEditor.Subtract(["NAME:Selections",

    "Blank Parts:=", str(objectString),

    "Tool Parts:=", str(bandTool)],

    ["NAME:SubtractParameters", "KeepOriginals:=",  False])



def RotateObject(name, skewAngle):

  oEditor.Rotate(["NAME:Selections",

	  "Selections:=", str(name),

    "NewPartsModelFlag:=", "Model"],

	  ["NAME:RotateParameters",

     "RotateAxis:=", "Z",

     "RotateAngle:=", str(skewAngle).replace(",", ".") + "deg"])



def SplitObject(name, side):

  oEditor.Split(["NAME:Selections",

    "Selections:=", str(name),

    "NewPartsModelFlag:=", "Model"],

    ["Name:SplitToParameters",

     "SplitPlane:=", "ZX",

     "WhichSide:=", str(side),

     "ToolType:=", "PlaneTool",

     "ToolEntityID:=", -1,

     "SplitCrossingObjectsOnly:=", False,

     "DeleteInvalidObjects:=", True])



def CreateRectangleObject(name, Xstart, Ystart, Zstart, Width, Height, Axis):

  oEditor.CreateRectangle(["NAME:RectangleParameters",

    "IsCovered:=", True,

    "XStart:=", str(Xstart),

    "YStart:=", str(Ystart),

    "ZStart:=", str(Zstart),

    "Width:=", str(Width),

    "Height:=", str(Height),

    "WhichAxis:=", str(Axis)],

    ["NAME:Attributes",

     "Name:=", str(name),

     "Flags:=", "",

     "Color:=", "(143 175 143)",

     "Transparency:=", 0,

     "PartCoordinateSystem:=", "Global",

     "UDMId:=", "",

     "MaterialValue:=", "\"\"",

     "SurfaceMaterialValue:=", "\"\"",

     "SolveInside:=", True,

     "ShellElement:=", False,

     "ShellElementThicknesS:=", "0mm",

     "ReferenceTemperature:=", "20cel",

     "IsMaterialEditable:=", True,

     "UseMaterialAppearance:=", False,

     "IsLightweight:=", False])



def SeparateCoil(coilName, R, G, B):

  oEditor.SeparateBody(["NAME:Selections",

	  "Selections:=", str(coilName),

	  "NewPartsModelFlag:=", "Model"],

	  ["CreateGroupsForNewObjects:=", False])

  ChangePropertyName(str(coilName), str(coilName) + "_1")

  ChangePropertyName(str(coilName) + "_Separate1", str(coilName) + "_2")
  oEditor.ChangeProperty(["NAME:AllTabs", ["NAME:Geometry3DAttributeTab",

    ["NAME:PropServers", str(coilName) + "_2"],

    ["NAME:ChangedProps",

    ["NAME:Color", "R:=", int(R), "G:=",  int(G), "B:=",  int(B)]]]])
```

## CreateMagnetCoordinateSystem

```python

def CreateMagnetCoordinateSystem(magnetName, origin_X, origin_Y, origin_Z, vector_XX, vector_XY, vector_YX, vector_YY, ModelType):

  oEditor.SetWCS(["NAME:SetWCS Parameter", "Working Coordinate System:=", "Global"])

  oEditor.CreateRelativeCS(["NAME:RelativeCSParameters",

    "Mode:=", "Axis/Position",

    "OriginX:=", str(origin_X).replace(",", ".") + "mm",

    "OriginY:=", str(origin_Y).replace(",", ".") + "mm",

    "OriginZ:=", str(origin_Z).replace(",", ".") + "mm",

    "XAxisXvec:=", str(vector_XX).replace(",", ".") + "mm",

    "XAxisYvec:=", str(vector_XY).replace(",", ".") + "mm",

    "XAxisZvec:=", "0mm",

    "YAxisXvec:=", str(vector_YX).replace(",", ".") + "mm",

    "YAxisYvec:=", str(vector_YY).replace(",", ".") + "mm",

    "YAxisZvec:=", "0mm"],

    ["NAME:Attributes", "Name:=", "CS_" + str(magnetName)])


  oEditor.ChangeProperty(["NAME:AllTabs",

    ["NAME:Geometry3DAttributeTab",

     ["NAME:PropServers", str(magnetName)],

     ["NAME:ChangedProps",

      ["NAME:Orientation", "Value:=", "CS_" + str(magnetName)]]]])
  




  if ModelType == 0:

    # Assign excitation to magnets for magnet eddy current loss calculations

    oBoundaryModule.AssignCurrent(["NAME:" + str(magnetName),

      "Objects:=", [str(magnetName)],

      "Current:=", "0A" ,

      "IsPositive:=", True])

```
